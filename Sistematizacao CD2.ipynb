```python
from pyspark.sql import SparkSession

df1 = spark.createDataFrame(df)
df1.head(10)
```

```python
# Coluna Age, 1 corresponde a 18-24 anos e 9 corresponde a 60-64 anos
# Quero remover os individuos com idade acima de 64 anos
df2 = df1.where(df1.Age.between(1, 9))
df2.head(10)
```

```python
from sklearn import datasets
from sklearn import linear_model
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.metrics import accuracy_score, classification_report
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.preprocessing import OrdinalEncoder

df4 = df3.toPandas()
x = df4.drop(columns=["Diabetes_binary", "ID", "MentHlth", "PhysHlth"])
y = df4['Diabetes_binary']

feat_continuas = ['BMI']
feat_ordinarias = ['Age', 'GenHlth', 'Income']
feat_binarias = ['HighBP', 'HighChol', 'CholCheck', 'Stroke', 'HeartDiseaseorAttack', 'Fruits', 'Veggies', 'HvyAlcoholConsump', 'AnyHealthcare', 'PhysActivity', 'Smoker']
preprocessar = ColumnTransformer(transformers=[('continuous', preprocessing.MinMaxScaler(), feat_continuas), ('binary', 'passthrough', feat_binarias), ('ordinal', OrdinalEncoder(), feat_ordinarias)])

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=50, stratify=y)

model = Pipeline([('preprocessar', preprocessar), ('logistic regression', LogisticRegression(max_iter=10000, class_weight={0: 1, 1: 3.5}, C=4))]) 
model.fit(x_train, y_train)
y_prob = model.predict_proba(x_test)[:, 1]
y_pred = y_prob > 0.5
print(classification_report(y_test, y_pred, target_names=['Sem Diabetes', 'Com Diabetes']))
print("Acurácia:", accuracy_score(y_test, y_pred))
```

```python
matriz_conf = metrics.confusion_matrix(y_test, y_pred)
sns.heatmap(matriz_conf, annot=True)
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier

model2 = DecisionTreeClassifier(max_depth=5, random_state=33, class_weight='balanced', min_samples_leaf=100)
model2.fit(x_train, y_train)
y_prob2 = model2.predict_proba(x_test)[:, 1]
y_pred2 = y_prob2 > 0.76
print(classification_report(y_test, y_pred2, target_names=['Sem Diabetes', 'Com Diabetes']))
print("Acurácia:", accuracy_score(y_test, y_pred2))
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(50,25))
plot_tree(model2, feature_names=x.columns, class_names=['Sem Diabetes', 'Com Diabetes'], filled=True, rounded=True, fontsize=10)
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier

model3 = RandomForestClassifier(n_estimators=300, max_depth=7, random_state=83, class_weight={0: 1, 1: 8}, min_samples_leaf=100, n_jobs=-1)
model3.fit(x_train, y_train)
y_prob3 = model3.predict_proba(x_test)[:, 1]
y_pred3 = y_prob3 > 0.6
print(classification_report(y_test, y_pred3, target_names=['Sem Diabetes', 'Com Diabetes']))
print("Acurácia:", accuracy_score(y_test, y_pred3))
```
